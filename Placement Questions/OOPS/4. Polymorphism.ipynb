## 4. Polymorphism




Polymorphism means one method can have different implementations in different classes.

## Example 1 : Different Classes, Same Method

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

## Example 2: Vehicle

```python

class Car:
    def start(self):
        print("Car starts with Key")

class Bike:
    def start(self):
        print("Bike starts with Kick")

c = Car()
b = Bike()

c.start()
b.start()
```

## Example: Birds

```python

class Sparrow:
    def fly(self):
        print("Sparrow flies")

class Eagle:
    def fly(self):
        print("Eagle flies high")

s = Sparrow()
e = Eagle()

s.fly()
e.fly()
```

## Example: Notifications


```python
class Email:
    def send(self):
        print("Email Sent")

class SMS:
    def send(self):
        print("SMS Sent")

e = Email()
s = SMS()

e.send()
s.send()
```

## Example: Payment Methods


```python
class CreditCard:
    def pay(self):
        print("Paid using Credit Card")

class Cash:
    def pay(self):
        print("Paid using Cash")

c = CreditCard()
k = Cash()

c.pay()
k.pay()
```
