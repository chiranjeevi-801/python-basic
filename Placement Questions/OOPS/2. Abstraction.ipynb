## 2. Abstraction:


Abstraction means showing only the essential details and hiding the implementation.


## Animal


```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
```

## Example: Payment System


```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass

class UPI(Payment):
    def pay(self):
        print("Paid using UPI")

u = UPI()
u.pay()
```

Example: Fan


```python
from abc import ABC, abstractmethod

class Fan(ABC):

    @abstractmethod
    def switch_on(self):
        pass

class CeilingFan(Fan):
    def switch_on(self):
        print("Fan is ON")

f = CeilingFan()
f.switch_on()
```

## Example: Printer


```python
from abc import ABC, abstractmethod

class Printer(ABC):

    @abstractmethod
    def print_doc(self):
        pass

class Inkjet(Printer):
    def print_doc(self):
        print("Printing Document")

p = Inkjet()
p.print_doc()
```
